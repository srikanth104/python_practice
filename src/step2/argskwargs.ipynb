Answer : items = [100, 200]  This creates a new list and makes items refer to it. It does not change what numbers refers to.

append() mutates the existing object.

= rebinds a variable to an object.

```python
def modify(items):
    items = [100, 200]  # Rebinds items to a new list
    items.append(300)    # Modifies that new list

numbers = [10, 20]
modify(numbers)

print(numbers)  # [10, 20]
```

The critical point is the order of operations.
1. Initially, items and numbers reference the same list.
2. items = [100, 200] makes items reference a new list.
3. items.append(300) modifies only that new list.
4. numbers still references the original [10, 20].

```python
a = [10, 20]
b = a

a[:] = [100, 200]

print(id(a))
print(id(b))
print(b)
print(a is b)
```

a and b still reference the same object, but its contents have changed.

Reassignment (=): Changes which object a variable references.
Mutation ([:] =, append(), clear()): Changes the existing list object.

() => Represents tuple in python

```python
def execute_agent(name, *tools):
    print(name)
    print(tools)

execute_agent(
    name="ResearchAgent",
    tools=("search", "calculator")
)

# *tools collects additional positional arguments, not keyword arguments.
# Python interprets tools= as a keyword argument. 
# But the function doesn't have a keyword parameter named tools, so it raises an error.
```

```python
def execute_agent(name, *tools):
    print(name)
    print(tools)

available_tools = ("search", "calculator")

execute_agent("ResearchAgent", *available_tools)

# Here, *available_tools unpacks the tuple into individual positional arguments.

#Syntax	            Meaning
#def f(*args)	    Collect positional arguments
#f(*values)	        Unpack an iterable into positional arguments
```
