```python
agents = [["Research"], ["Code"]]

backup = agents.copy()

print(id(agents))
print(id(backup))


print(id(agents[0]))
print(id(backup[0]))
print("______________________")
print(id(backup))
backup[0].append("Planner") # This modifies shared nested list in the both places 
print(id(backup))

backup.append(["Review"]) # This add the element at the end for only backup 
print(id(backup))
print(id(backup[2]))

backup[0] = ["Planner"]  # We are modifying the existing reference doesn't change agent


print(agents)
print(backup)
print(agents is backup)
print(agents[0] is backup[0])
```

```python
from collections import deque
items = deque([10,20,30])

print(type(items))
# Deque allows elements from both the ends 

items.append(40)
items.appendleft("Hello")
items.insert(0,1)
items.pop()
items.popleft()

print(items)
```

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a - b)
print(b - a)
print(a & b)

# Operation	    Meaning	                    Result
# a - b	    In A, not in B	                {1, 2}
# b - a	    In B, not in A	                {5, 6}
# a & b	    Common to both	                {3, 4}
```

```python
a = (1, 2)
b = (1, 2)

print(a == b)
print(hash(a) == hash(b))

# a == b → True, because both tuples contain the same elements.
# hash(a) == hash(b) → True, because equal objects must have equal hash values.
```
