```python
# 1. Create a tuple. Python builds a "box" in memory at address A.
my_tuple = (1, 2, 3)
print(id(my_tuple))  # Output: 1402394829344 (Address A)

# 2. Try to change an item inside Address A
# my_tuple = 99 <-- ❌ Python blocks this because Address A is IMMUTABLE.

# 3. Reassign the variable. Python builds a brand-new "box" at address B.
my_tuple = (1, 2, 99)
print(id(my_tuple))  # Output: 1402394829568 (Address B - A completely different 
```
