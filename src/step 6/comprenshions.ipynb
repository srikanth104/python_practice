```python
# 	                             Generator expression	            Generator function
# Syntax	                        (expr for x in items)	            def with yield
# Best for	                    Simple transform or filter	        Multi-step logic
# Multiple statements	            No, a single expression	            Yes, any code
# Keeping state between yields	Not really	                        Yes, with local variables
# Try/except, setup, cleanup	    No	                                Yes
# Reusable and parameterized	    Not by itself	                    Yes, call it again for a fresh generator
# Readability	                    Compact for simple cases	        Clearer for complex cases
```

```python
agents = ["Research", "Code", "Review"]

result1 = {}

for agent in agents:
    result1[agent] = agent.upper()

result2 = {
    agent: agent.upper()
    for agent in agents
}
    
print(result1)
print(result2)
```

Next(iterator) will print 10 and for loop resume from there 

```python
# iter(numbers) creates an iterator.
# next(iterator) prints 10 and advances that iterator.
# for n in numbers creates a new iterator from the original list.
# The new iterator starts at the beginning and prints 10, 20, 30.

numbers = [10, 20, 30]
iterator = iter(numbers)

print(next(iterator))

for n in numbers:
    print(n)
    
```

```python
def process(n):
    print(f"Processing {n}")
    return n * 10

gen = (process(n) for n in [1,2,3])

print("Generator Created")
print(next(gen))

# generator is lazy loading
```

```python
def get_tasks(): 
    yield "Research"
    yield "Code"
    yield "Review"

tasks = get_tasks()

task_count = sum(1 for task in tasks)

print("Count: ", task_count)

for task in tasks:
    print("Executing: ", task)
```

```python
numbers = [1, 2, 3]

gen = (n * 10 for n in numbers)

numbers.append(4)

print(list(gen))

# gen is created, but its values haven't been calculated.
# numbers.append(4) modifies the original list.
# list(gen) consumes the generator.
# The generator sees the appended element and produces 40 as well.
```

ZIP - Combines multiple iterators and one position at a time

```python
agents = ["Research", "Code", "Review"]
ids = [101, 102, 103]

results = zip(agents, ids)
print(list(results))
print(type(results))

# Zip returns an iterator not a list
```

```python
agents = ["Research", "Code", "Review"]
ids = [101, 102]

results = zip(agents, ids)
print(list(results))
# This behavior can silently discard data when two collections have different lengths.
```

```python
agents = ["Research", "Code", "Review"]
ids = [101, 102]

results = zip(agents, ids, strict=True)
print(list(results))
```

enumerate => is another built-in python funciton that returns an iterator

```python
agents = ["Research", "Code", "Review"]
for index,agent in enumerate(agents):
    print(index, agent)
```

```python
agents = ["Research", "Code", "Review"]
for index,agent in enumerate(agents, start=1):
    print(index, agent)
```

```python
# enumerate add an index to each item in an interable 
# Zip pairs up items for several iterable position by position. 
```

```python
agents = {"Research", "Code", "Review"}
ids = [101,102,103]

for index, (agents, ids) in enumerate(zip(agents, ids), start=1):
    print(index, agents, ids)
    
```

```python
agents = {"Research", "Code", "Review"}
ids = [101,102,103]

result = list(enumerate(zip(agents, ids),start=1))
print(result)

# First ZIP produces the tuple 
# Then enumerate assigns index starting from 1 
# enumerate doesn't flattern nested tuples, it creates a new outer tuple containing the index and original value
```

```python
# filter() and Lazy Evaluation 

numbers = [1, 2, 3, 4, 5, 6]
result = filter(lambda n: n %2 == 0, numnbers)
print(list(result))
```

```python
def is_even(n):
    print(f"Checking {n}")
    return n %2 == 0

numnbers = [1,2,3,4]
result = filter(is_even, numnbers)  # It creates a filter but doesn't execute

print("Filter Created")
print(next(result))
```

```python
def generate():
    try:
        yield 1
        yield 2
    finally:
        print("Cleanup")

gen = generate()

print(next(gen))
print(next(gen))
gen.close()
print("Finished")

# Gen.close will initiate cleanup process 
```
