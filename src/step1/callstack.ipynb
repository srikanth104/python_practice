```python
#  What happens to the call stack?
# This is the important part.

# When a generator executes yield:
# 1. It produces a value for the caller.
# 2. Its execution is suspended.
# 3. Its local variables and execution position are preserved.
# 4. Its invocation is no longer actively executing on the call stack.
# 5. A later next() can resume it from where it stopped.

#yield suspends execution; it doesn't destroy the generator's execution state.


def normal():
    print("Started")
    return 10

def generator():
    print("Started")
    yield 10
    print("Resumed")
    yield 20
print(normal())
print("Yield Execcution")
g = generator()
print(next(g))
print(next(g))
```

```python
def counter():
    x = 10
    yield x

    x += 5
    yield x

g = counter()

print(next(g))  # 10
print(next(g))  # 15
print(next(g))  # StopIteration  
```

```python
# First next(g) executes yield 10, returning 10 and suspending the generator.
# Second next(g) resumes at return 20.
# return terminates the generator immediately.
# yield 30 is never executed.

def numbers():
    yield 10
    return 20
    yield 30

g = numbers()

print(next(g))
print(next(g))
```

```python
# list() collects values produced by yield, not the generator's final return value.
# - yield 1 → adds 1 to the list.
# - yield 2 → adds 2 to the list.
# - return 3 → terminates the generator, raising StopIteration with value = 3.
# - list() handles StopIteration and discards its return value.

# Remember: yield produces iteration values; return ends the generator.

def example():
    yield 1
    yield 2
    return 3

print(list(example()))
```

```python
def example():
    yield 1
    yield 2
    return 3

g = example()

print(next(g))
print(next(g))

try:
    next(g)
except StopIteration as e:
    print(e.value)
```

```python
def inner():
    print("Inner started")
    yield 10
    yield 20
    print("Inner finished")

def outer():
    print("Outer started")
    yield from inner()
    print("Outer finished")

g = outer()
print(next(g))
print(next(g))
```

```python
# Call	Execution	Result
# next(g) #1	Starts outer() and inner()	Yields 10
# next(g) #2	Resumes inner()	Yields 20
# next(g) #3	Finishes inner(), then outer()	Raises StopIteration

def inner():
    print("Inner started")
    yield 10
    yield 20
    print("Inner finished")

def outer():
    print("Outer started")
    yield from inner()
    print("Outer finished")

g = outer()
print(next(g))
print(next(g))
print(next(g))
```

```python
# parent() delegates execution to child() using yield from.
# child() yields 1, which reaches list().
# child() executes return 5, terminating the child generator.
# yield from captures the return value 5 and assigns it to x.
# parent() executes yield x * 2, producing 10.

def child():
    yield 1
    return 5

def parent():
    x = yield from child()
    yield x * 2

print(list(parent()))
```

Important: send(value) resumes the generator and supplies value as the result of its currently suspended yield expression.

next(g) is effectively equivalent to g.send(None) for a generator.

```python
# next(g) suspended execution at yield 10.
# g.throw(ValueError(...)) injected an exception at that suspended yield.
# The generator's except ValueError handled it.
# Execution continued to yield 20.

#If the generator doesn't handle the exception, it propagates back to the caller.

def worker():
    try:
        yield 10
    except ValueError:
        print("Worker caught ValueError")

    yield 20

g = worker()

print(next(g))
print(g.throw(ValueError("Something failed")))
```

```python
# The exception is sent to the parent, but the active yield from delegation forwards it to the child.
# The child handles the exception and continues execution.
# There are edge cases: for example, if the child cannot handle the exception, it propagates through the parent. GeneratorExit also has special handling during generator closure.

def child():
    try:
        yield 10
    except ValueError:
        print("Child handled exception")

    yield 20

def parent():
    yield from child()

g = parent()

print(next(g))
print(g.throw(ValueError("failed")))
```
