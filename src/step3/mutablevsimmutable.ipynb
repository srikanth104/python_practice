.copy() creates a new list object containing references to the same elements.
list.copy is a shallow copy.

```python
a = [1.2]
b = a.copy()

print(id(a))
print(id(b))

b.append(3)
print(id(b))
print(id(a))
print(a)
print(b)
```

```python
config = {
    "model": "gpt-4.1",
    "settings": {
        "temperature": 0.5
    }
}

new_config = config.copy()

new_config["model"] = "gpt-4o"
new_config["settings"]["temperature"] = 0.9

print(config)
print(new_config)

#A shallow copy creates a new outer container but preserves references to its elements.
```

```python
config = {
    "model": "gpt-4.1",
    "settings": {
        "temperature": 0.5
    }
}

new_config = config.copy()

new_config["settings"] = {
    "temperature": 0.9
}

print(config["settings"]["temperature"])
print(new_config["settings"]["temperature"])
```

```python
# copy() is a shallow copy 
# copy.deepcopy() A Shallow copy duplicates only the outer container. A deep copy recursvily 
# copies nested objects

import copy
a = {"tools" : ["search", "calculator"]}

c = a

b = copy.deepcopy(a)

print(id(a))
print(id(c))
print("_________________")
print(id(b))
print(b)

print("_________________")
b = {"tools" : ["python"]}
print(a)
print(b)
```

```python
import copy

a = {
    "tools": ["search", "calculator"]
}

b = copy.deepcopy(a)
print("_________________")
print(id(a))
print(id(b))
print("_________________")

b["tools"].append("browser")
print(id(a))
print(id(b))
print(a["tools"])
print(b["tools"])
```
