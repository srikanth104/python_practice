```python
def numbers(): 
    for i in range(3):
        print("Before", i)
        yield i
        print("After", i)
g = numbers()

print(next(g))
print(next(g)) 
```

```python
def child():
    yield 10
    yield 20 
    return 30

def parent():
    result = yield from child()
    print("Child returned:", result)
    yield 30

g = parent()

for value in g:
    print(value)
    

```

generators are not automatically faster. They trade eager materialization for incremental computation, which can reduce peak memory but introduce per-item iteration overhead.

```python
def generate():
    try:
        yield 10 
    finally:
        yield 20 

g = generate()

print(next(g))
g.close  # Raises an exception if we use yiedl in finally
```

```python
def generate():
    yield 10
    yield 20

g = generate()

print(next(g))

try:
    g.throw(ValueError("Failed"))
except ValueError:
    print("Caught outside")

print(next(g)) 
```

```python
def child():
    x = yield 10
    print("Child received:", x)
    yield 20

def parent():
    yield from child()
    yield 30

g = parent()

print(next(g))
print(g.send(50))
print(next(g))
```

```python
import inspect

def generate():
    yield 10
    raise ValueError("Failed")
    yield 20

g = generate()

print(next(g))

try:
    next(g)
except ValueError:
    print("Caught")

print(inspect.getgeneratorstate(g))
```
